```python
import re
import pandas as pd

# Load the HTML file
with open("uhi_heatmap.html", "r", encoding="utf-8") as file:
    html_content = file.read()

# Use regex to extract all coordinate-intensity triplets
matches = re.findall(r"\[\s*(-?\d+\.\d+),\s*(-?\d+\.\d+),\s*(-?\d+\.\d+)\s*\]", html_content)

if matches:
    # Convert matches into a DataFrame
    heatmap_list = [[float(lat), float(lon), float(intensity)] for lat, lon, intensity in matches]
    heatmap_df = pd.DataFrame(heatmap_list, columns=["Latitude", "Longitude", "Heatmap_Intensity"])

    # Save to CSV
    heatmap_df.to_csv("heatmap_extracted.csv", index=False)
    print(f"✅ Extracted {len(heatmap_df)} heatmap points and saved to 'heatmap_extracted.csv'")
else:
    print("⚠️ Heatmap data not found in the HTML file.")
```
